# 00 - Download data

Downloads the UCI Online Retail II dataset into `data/raw/`.

Source: https://archive.ics.uci.edu/dataset/502/online+retail+ii

If the download fails, get the zip manually from the page above and extract it into `data/raw/`.

```python
import io
import urllib.request
import zipfile
from pathlib import Path

URL = "https://archive.ics.uci.edu/static/public/502/online+retail+ii.zip"
RAW_DIR = Path.cwd().parent / "data" / "raw"  # notebook runs from notebooks/
RAW_DIR.mkdir(parents=True, exist_ok=True)
RAW_DIR
```

```python
print(f"Downloading {URL} ...")
with urllib.request.urlopen(URL) as resp:
    payload = resp.read()

with zipfile.ZipFile(io.BytesIO(payload)) as zf:
    zf.extractall(RAW_DIR)
    print("Extracted:", ", ".join(zf.namelist()))
```

```python
for f in sorted(RAW_DIR.iterdir()):
    print(f"{f.name:40} {f.stat().st_size / 1e6:8.1f} MB")
```
